### GROQ MODEL INTEGRATION


```python
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

load_dotenv()

model=init_chat_model("qwen/qwen3.8-27b", model_provider="groq")
response=model.invoke("Who is the captain of indian cricket team in test")
response.content
```

### STREAMING

### checking without streaming

```python

model.invoke("write a 200 words paragraph about Indian Captain Rohit Sharma")
```

### with streaming

```python
for chunk in model.stream("write a 200 words paragraph about Indian Captain Rohit Sharma"):
    print(chunk.text)
```

```python
for chunk in model.stream("write a paragraph about Indian Captain Rohit Sharma"):
    print(chunk.text, end="|", flush= True)
```

```python
for chunk in model.stream("Why do parrots have colorful feathers"):
    print(chunk.text, end="|", flush= True)
```

### BATCHING


```python
responses = model.batch([
    "Master of PULL shot in cricket?",
    "Rohit Sharma Wife name?",
    "Against Which team rohit sharma make his latest century in ODIS?"
])
for response in responses:
    print(response)
```
